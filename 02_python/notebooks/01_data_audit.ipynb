## 5. Count Variable Validation

This section checks whether rental counts are non-negative and verifies that total rentals equal the sum of casual and registered rentals.

```python
count_validation = pd.DataFrame({
    'dataset': ['day', 'hour'],

    'negative_casual': [
        (day['casual'] < 0).sum(),
        (hour['casual'] < 0).sum()
    ],

    'negative_registered': [
        (day['registered'] < 0).sum(),
        (hour['registered'] < 0).sum()
    ],

    'negative_cnt': [
        (day['cnt'] < 0).sum(),
        (hour['cnt'] < 0).sum()
    ],

    'cnt_mismatch': [
        (day['cnt'] != day['casual'] + day['registered']).sum(),
        (hour['cnt'] != hour['casual'] + hour['registered']).sum()
    ]
})

display(count_validation)
```

### Count validation result

No negative rental counts were found. In both datasets, total rentals (`cnt`) are equal to the sum of casual and registered rentals in every record.

No correction is required for the count variables.

## 7. Calendar Consistency Validation
This section checks whether the recorded year, month, and weekday values are consistent with the date column.

```python
# calculate expected calendar values from the data column

day_expected_year = day_dates.dt.year -2011
hour_expected_year = hour_dates.dt.year -2011

day_expected_month = day_dates.dt.month
hour_expected_month = hour_dates.dt.month


day_expected_weekday = (day_dates.dt.dayofweek + 1) %7
hour_expected_weekday = (hour_dates.dt.dayofweek + 1) %7


```

```python
calendar_validation = pd.DataFrame({
    'dataset': ['day', 'hour'],

    'year_mismatch': [
        (day['yr'] != day_expected_year).sum(),
        (hour['yr'] != hour_expected_year).sum()
    ],

    'month_mismatch': [
        (day['mnth'] != day_expected_month).sum(),
        (hour['mnth'] != hour_expected_month).sum()
    ],

    'weekday_mismatch': [
        (day['weekday'] != day_expected_weekday).sum(),
        (hour['weekday'] != hour_expected_weekday).sum()
    ]
})

display(calendar_validation)
```

### Calendar consistency result

The recorded year, month, and weekday values are fully consistent with the date column in both datasets.

No calendar inconsistencies were identified.

```python
day_expected_workingday = (
    (day['holiday'] == 0) &
    (~day['weekday'].isin([0, 6]))
).astype(int)

hour_expected_workingday = (
    (hour['holiday'] == 0) &
    (~hour['weekday'].isin([0, 6]))
).astype(int)

day_workingday_mismatch = (
    day['workingday'] != day_expected_workingday
).sum()

hour_workingday_mismatch = (
    hour['workingday'] != hour_expected_workingday
).sum()

print(
    'workingday mismatches in day dataset:',
    day_workingday_mismatch
)

print(
    'workingday mismatches in hour dataset:',
    hour_workingday_mismatch
)
```

### Working-day validation result
The working-day variable is consistent with the holiday and weekday variables in both datasets.
No working-day inconsistencies were identified.

## 8. Hourly Record Completeness
This section checks how many hourly records are available for each date and identifies dates with fewer than 24 hourly observations.


```python
hour_records_per_day = (
    hour.groupby('dteday').size()
    .reset_index(name= 'hour_record_count')
)
display (hour_records_per_day.head())
```

```python
complete_days = (
    hour_records_per_day['hour_record_count'] == 24
).sum()

incomplete_days = (
    hour_records_per_day['hour_record_count'] < 24
).sum()

print('dates with 24 hourly records:', complete_days)
print('dates with fewer than 24 hourly records:', incomplete_days)
```

```python
incomplete_hourly_dates = hour_records_per_day[
    hour_records_per_day['hour_record_count'] < 24
].sort_values('hour_record_count')

display(incomplete_hourly_dates.head(20))
```

### Hourly completeness result

The hourly dataset contains 655 complete dates with 24 hourly records and 76 incomplete dates with fewer than 24 records.

This represents structural time gaps rather than cell-level missing values. No missing hourly records will be imputed during the raw data audit.

```python
expected_hour_records = len(hour_records_per_day) * 24
observed_hour_records = len(hour)

missing_hour_records = (
    expected_hour_records - observed_hour_records
)

print('expected hourly records:', expected_hour_records)
print('observed hourly records:', observed_hour_records)
print('missing hourly records:', missing_hour_records)
```

```python
expected_hours = set(range(24))

missing_hours_by_date = (
    hour.groupby('dteday')['hr']
    .apply(
        lambda hours: sorted(
            expected_hours - set(hours)
        )
    )
    .reset_index(name='missing_hours')
)

missing_hours_by_date['missing_hour_count'] = (
    missing_hours_by_date['missing_hours']
    .apply(len)
)

missing_hours_by_date = missing_hours_by_date[
    missing_hours_by_date['missing_hour_count'] > 0
].sort_values(
    'missing_hour_count',
    ascending=False
)

display(missing_hours_by_date.head(20))
```

### Hourly completeness finding

The hourly dataset contains 17,379 observed records, compared with 17,544 expected date-hour combinations. Therefore, 165 hourly records are absent across 76 dates.

The gaps are structural because the corresponding rows do not exist in the dataset. They are not cell-level missing values. No hourly records will be imputed or removed during the raw data audit.

```python
issue_register = pd.DataFrame({
    'issue_id': ['dq-01'],
    'dataset': ['hour'],
    'issue': ['missing date-hour records'],
    'description': [
        '165 expected date-hour combinations are absent across 76 dates.'
    ],
    'severity': ['medium'],
    'decision': [
        'retain the raw data and evaluate the effect before hourly analysis'
    ]
})

display(issue_register)
```

## 9. Daily and Hourly Demand Reconciliation

This section aggregates hourly rental counts by date and compares them with the corresponding daily totals.

```python
hourly_daily_totals = (
    hour.groupby('dteday', as_index=False)[
        ['casual', 'registered', 'cnt']
    ]
    .sum()
    .rename(columns={
        'casual': 'hour_casual',
        'registered': 'hour_registered',
        'cnt': 'hour_cnt'
    })
)

daily_totals = (
    day[
        ['dteday', 'casual', 'registered', 'cnt']
    ]
    .rename(columns={
        'casual': 'day_casual',
        'registered': 'day_registered',
        'cnt': 'day_cnt'
    })
)
```

```python
daily_hourly_check = daily_totals.merge(
    hourly_daily_totals,
    on='dteday',
    how='left'
)

daily_hourly_check['casual_mismatch'] = (
    daily_hourly_check['day_casual'] !=
    daily_hourly_check['hour_casual']
)

daily_hourly_check['registered_mismatch'] = (
    daily_hourly_check['day_registered'] !=
    daily_hourly_check['hour_registered']
)

daily_hourly_check['cnt_mismatch'] = (
    daily_hourly_check['day_cnt'] !=
    daily_hourly_check['hour_cnt']
)
```

```python
demand_reconciliation_summary = pd.DataFrame({
    'variable': ['casual', 'registered', 'cnt'],
    'mismatch_dates': [
        daily_hourly_check['casual_mismatch'].sum(),
        daily_hourly_check['registered_mismatch'].sum(),
        daily_hourly_check['cnt_mismatch'].sum()
    ]
})

display(demand_reconciliation_summary)
```

### Daily-hourly reconciliation result

The aggregated hourly rental counts exactly match the corresponding daily totals for casual, registered, and total rentals.
No demand inconsistencies were identified between the daily and hourly datasets. However, the previously identified structural hourly gaps remain relevant for hour-level analysis.

## 10. Weather Variable Range Validation

This section checks the minimum and maximum values of the normalized weather variables and identifies values outside the expected range of 0 to 1.

```python
weather_columns = [
    'temp',
    'atemp',
    'hum',
    'windspeed'
]
day_weather_range = day[weather_columns].agg(
    ['min', 'max']
).T
hour_weather_range = hour[weather_columns].agg(
    ['min', 'max']
).T
print('weather variable ranges in day dataset:')
display(day_weather_range)
print('weather variable ranges in hour dataset:')
display(hour_weather_range)

```

```python
day_weather_out_of_range = (
    (day[weather_columns] < 0) |
    (day[weather_columns] > 1)
).sum()

hour_weather_out_of_range = (
    (hour[weather_columns] < 0) |
    (hour[weather_columns] > 1)
).sum()

weather_range_validation = pd.DataFrame({
    'variable': weather_columns,
    'day_out_of_range': [
        day_weather_out_of_range[column]
        for column in weather_columns
    ],
    'hour_out_of_range': [
        hour_weather_out_of_range[column]
        for column in weather_columns
    ]
})

display(weather_range_validation)
```

### Weather range validation result
All normalized weather variables in both datasets fall within the expected range of 0 to 1.
No out-of-range weather values were identified.

## 11. Weather Boundary Value Review

This section counts exact zero values in the normalized weather variables and reviews potentially unusual boundary observations.

```python
day_weather_zero_counts = (
    day[weather_columns] == 0
).sum()

hour_weather_zero_counts = (
    hour[weather_columns] == 0
).sum()

weather_zero_summary = pd.DataFrame({
    'variable': weather_columns,
    'day_zero_count': [
        day_weather_zero_counts[column]
        for column in weather_columns
    ],
    'hour_zero_count': [
        hour_weather_zero_counts[column]
        for column in weather_columns
    ]
})

display(weather_zero_summary)
```

```python
print('daily records with zero humidity:')
display(
    day.loc[
        day['hum'] == 0,
        ['dteday', 'season', 'weathersit', 'temp', 'hum', 'windspeed', 'cnt']
    ]
)

print('hourly records with zero humidity:')
display(
    hour.loc[
        hour['hum'] == 0,
        ['dteday', 'hr', 'season', 'weathersit', 'temp', 'hum', 'windspeed', 'cnt']
    ]
)

print('hourly records with zero apparent temperature:')
display(
    hour.loc[
        hour['atemp'] == 0,
        ['dteday', 'hr', 'temp', 'atemp', 'hum', 'windspeed', 'cnt']
    ]
)
```

### Weather boundary review conclusion

The weather boundary-value review identified the following observations:

- Zero humidity is concentrated on March 10, 2011, with one daily record and 22 hourly records. This unusual pattern has been recorded as issue `dq-02`.
- Two zero apparent-temperature values occur on January 22, 2011. Their corresponding temperature values are also close to zero, so they are considered plausible boundary observations.
- Zero wind-speed values remain within the valid normalized range and are not treated as missing values.

All original values are retained without modification during the raw data audit.

```python
new_issue = pd.DataFrame({
    'issue_id': ['dq-02'],
    'dataset': ['day and hour'],
    'issue': ['zero humidity values'],
    'description': [
        'Humidity is recorded as zero on 2011-03-10 in one daily record and 22 hourly records.'
    ],
    'severity': ['medium'],
    'decision': [
        'retain the original values and evaluate their effect during later analysis'
    ]
})

if 'dq-02' not in issue_register['issue_id'].values:
    issue_register = pd.concat(
        [issue_register, new_issue],
        ignore_index=True
    )

display(issue_register)
```

## 12. Identifier Sequence Validation

```python
day_expected_instant = pd.Series(
    range(1, len(day) + 1),
    index=day.index
)

hour_expected_instant = pd.Series(
    range(1, len(hour) + 1),
    index=hour.index
)

instant_validation = pd.DataFrame({
    'dataset': ['day', 'hour'],

    'minimum_instant': [
        day['instant'].min(),
        hour['instant'].min()
    ],

    'maximum_instant': [
        day['instant'].max(),
        hour['instant'].max()
    ],

    'expected_maximum': [
        len(day),
        len(hour)
    ],

    'sequence_mismatch': [
        (day['instant'] != day_expected_instant).sum(),
        (hour['instant'] != hour_expected_instant).sum()
    ]
})

display(instant_validation)
```

## 13. Cross-Dataset Calendar Consistency

This section checks whether the calendar-related variables recorded in the hourly dataset are consistent with the corresponding daily records.

```python
calendar_columns = [
    'season',
    'yr',
    'mnth',
    'holiday',
    'weekday',
    'workingday'
]

hour_calendar_unique_counts = (
    hour.groupby('dteday')[calendar_columns]
    .nunique()
)

hour_calendar_inconsistencies = (
    hour_calendar_unique_counts > 1
).sum()

display(
    hour_calendar_inconsistencies
    .rename('inconsistent_dates')
    .to_frame()
)
```

```python
hour_calendar_daily = (
    hour.groupby('dteday', as_index=False)[calendar_columns]
    .first()
)

display(hour_calendar_daily.head())
```

```python
day_calendar = day[
    ['dteday'] + calendar_columns
].copy()
```

```python
calendar_cross_check = day_calendar.merge(
    hour_calendar_daily,
    on='dteday',
    how='left',
    suffixes=('_day', '_hour')
)

display(calendar_cross_check.head())
```

```python
cross_dataset_calendar_summary = pd.DataFrame({
    'variable': calendar_columns,

    'mismatch_dates': [
        (
            calendar_cross_check[f'{column}_day'] !=
            calendar_cross_check[f'{column}_hour']
        ).sum()
        for column in calendar_columns
    ]
})

display(cross_dataset_calendar_summary)
```

### Cross-dataset calendar consistency result

The calendar-related variables are internally consistent within the hourly dataset and fully match the corresponding daily records.

No cross-dataset calendar inconsistencies were identified.

## 14. Daily and Hourly Weather Consistency

This section compares the daily weather values with the mean hourly weather values for each date.

```python
weather_columns = [
    'temp',
    'atemp',
    'hum',
    'windspeed'
]

hourly_daily_weather = (
    hour.groupby('dteday', as_index=False)[weather_columns]
    .mean()
    .rename(columns={
        'temp': 'hour_temp_mean',
        'atemp': 'hour_atemp_mean',
        'hum': 'hour_hum_mean',
        'windspeed': 'hour_windspeed_mean'
    })
)

display(hourly_daily_weather.head())
```

```python
daily_weather = day[
    ['dteday'] + weather_columns
].rename(columns={
    'temp': 'day_temp',
    'atemp': 'day_atemp',
    'hum': 'day_hum',
    'windspeed': 'day_windspeed'
})

weather_cross_check = daily_weather.merge(
    hourly_daily_weather,
    on='dteday',
    how='left'
)

display(weather_cross_check.head())
```

```python
weather_difference_summary = pd.DataFrame({
    'variable': weather_columns,

    'maximum_absolute_difference': [
        (
            weather_cross_check['day_temp'] -
            weather_cross_check['hour_temp_mean']
        ).abs().max(),

        (
            weather_cross_check['day_atemp'] -
            weather_cross_check['hour_atemp_mean']
        ).abs().max(),

        (
            weather_cross_check['day_hum'] -
            weather_cross_check['hour_hum_mean']
        ).abs().max(),

        (
            weather_cross_check['day_windspeed'] -
            weather_cross_check['hour_windspeed_mean']
        ).abs().max()
    ]
})

display(weather_difference_summary)
```

### Daily-hourly weather consistency result

The daily temperature, apparent temperature, humidity, and wind-speed values closely match the corresponding mean hourly values.

The maximum absolute differences are below 0.000001 and are attributable to numerical rounding. No meaningful weather inconsistencies were identified between the two datasets.
